```python
filtered_bact="/content/drive/MyDrive/Subtractive Genomics Approach/m_tuberculosis_filtered.fasta"
```

```python
with open(filtered_bact, "r") as handle:
    bact_dict = SeqIO.to_dict(SeqIO.parse(handle, "fasta"))
bact_dict
```

```python
bact_series = pd.Series(bact_dict)
bact_series
```

```python
clean_series = bact_series.drop(homo_ids)
```
